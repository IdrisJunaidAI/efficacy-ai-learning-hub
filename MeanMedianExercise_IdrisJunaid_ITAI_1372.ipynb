# Exercise: Mean & Median Customer Spend

Here's some code that will generate some random e-commerce data; just an array of total amount spent per transaction. Select the code block, and hit "play" to execute it:

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

incomes = np.random.normal(100.0, 20.0, 10000)

plt.hist(incomes, 50)
plt.show()
```

Now, find the mean and median of this data. In the code block below, write your code, and see if your result makes sense:

```python
# Calculate the mean and median of the generated customer spending data
mean_income = np.mean(incomes)
median_income = np.median(incomes)

print("Mean:", mean_income)
print("Median:", median_income)
```

This is pretty much the world's easiest assignment, but we're just trying to get your hands on iPython and writing code with numpy to get you comfortable with it.

Try playing with the code above to generate different distributions of data, or add outliers to it to see their effect.

```python
# Add an outlier and compare the mean and median
incomes_with_outlier = np.append(incomes, 1000)

print("Mean with outlier:", np.mean(incomes_with_outlier))
print("Median with outlier:", np.median(incomes_with_outlier))
```
